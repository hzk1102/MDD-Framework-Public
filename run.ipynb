```python
import os
import sys
import pandas as pd
import soundfile as sf
import torch
import torch.nn.functional as F
from transformers import Wav2Vec2Processor, Wav2Vec2ForCTC, HubertForCTC, Wav2Vec2FeatureExtractor

MDD_REPO = "/home/h/git/MDD-Framework-Public"
sys.path.insert(0, MDD_REPO)

from model import PL
from utils import (
    SAMPLE_RATE, BLANK_TOKEN_ID, PAD_TOKEN_ID,
    build_feature_extractor, load_vocab, text_to_tensor,
)

# ── Config ────────────────────────────────────────────────────
BASE_DIR        = "/home/h/Data/MDD-Challenge-2025-private-test/MDD-Challenge-2025-private-test"
CSV_PATH        = "/home/h/Data/MDD-Challenge-2025-private-test/MDD-Challenge-2025-private-test/metadata/private_test_submission.csv"
VOCAB_PATH      = os.path.join(MDD_REPO, "vocab_extended.json")
CHECKPOINT_PATH = os.path.join(MDD_REPO, "checkpoint/checkpoint_wl.pth")
PRETRAINED_BASE = "nguyenvulebinh/wav2vec2-base-vietnamese-250h"
OUTPUT_CSV      = "results.csv"

device = "cuda" if torch.cuda.is_available() else "cpu"

# ── Model 1: wav2vec2 Vietnamese ──────────────────────────────
processor1 = Wav2Vec2Processor.from_pretrained(PRETRAINED_BASE)
model1 = Wav2Vec2ForCTC.from_pretrained(PRETRAINED_BASE).to(device).eval()

# ── Model 2: HuBERT English (chỉ dùng confidence, không decode text) ──
feature_extractor2 = Wav2Vec2FeatureExtractor.from_pretrained("facebook/hubert-base-ls960")
model2 = HubertForCTC.from_pretrained("facebook/hubert-base-ls960").to(device).eval()

# ── Model 3: PL checkpoint ────────────────────────────────────
vocab      = load_vocab(VOCAB_PATH)
vocab_size = len(vocab)
id2token   = {idx: tok for tok, idx in vocab.items()}
feature_extractor3 = build_feature_extractor()

model3 = PL.from_pretrained(PRETRAINED_BASE, vocab_size=vocab_size)
state_dict = torch.load(CHECKPOINT_PATH, map_location=device)
state_dict = {k.replace("module.", ""): v for k, v in state_dict.items()}
missing, unexpected = model3.load_state_dict(state_dict, strict=False)
if missing:    print(f"[WARN] Missing    ({len(missing)}): {missing[:3]}")
if unexpected: print(f"[WARN] Unexpected ({len(unexpected)}): {unexpected[:3]}")
model3 = model3.to(device).eval()
print("Loaded checkpoint_wl.pth OK")


# ── Helpers ───────────────────────────────────────────────────
def get_trans_conf_m1(speech):
    """Model1: Wav2Vec2ForCTC Vietnamese — trả về (transcription, confidence)."""
    inputs = processor1(speech, sampling_rate=16000, return_tensors="pt", padding=True)
    input_values = inputs.input_values.to(device)
    with torch.no_grad():
        logits = model1(input_values).logits
    log_probs  = F.log_softmax(logits, dim=-1)
    pred_ids   = torch.argmax(logits, dim=-1)
    confidence = log_probs[0, torch.arange(logits.shape[1]), pred_ids[0]].mean().item()
    transcription = processor1.batch_decode(pred_ids)[0]
    return transcription, confidence


def get_conf_m2(speech):
    """Model2: HuBERT English — chỉ trả về confidence (không decode text tiếng Việt)."""
    inputs = feature_extractor2(speech, sampling_rate=16000, return_tensors="pt", padding=True)
    input_values = inputs.input_values.to(device)
    with torch.no_grad():
        logits = model2(input_values).logits
    log_probs  = F.log_softmax(logits, dim=-1)
    pred_ids   = torch.argmax(logits, dim=-1)
    confidence = log_probs[0, torch.arange(logits.shape[1]), pred_ids[0]].mean().item()
    return confidence


def get_trans_conf_m3(speech, canonical_text):
    """Model3: PL checkpoint — trả về (transcription, confidence)."""
    inputs = feature_extractor3(
        [speech], sampling_rate=SAMPLE_RATE, padding=True, return_tensors="pt"
    )
    input_values = inputs.input_values.to(device)

    token_ids  = text_to_tensor(canonical_text, vocab) or [PAD_TOKEN_ID]
    linguistic = torch.tensor(token_ids, dtype=torch.long, device=device).unsqueeze(0)

    with torch.no_grad():
        logits = model3(input_values, linguistic)

    log_probs  = F.log_softmax(logits, dim=-1)
    pred_ids   = torch.argmax(logits, dim=-1)
    confidence = log_probs[0, torch.arange(logits.shape[1]), pred_ids[0]].mean().item()

    # Greedy CTC decode
    collapsed, prev = [], None
    for tid in pred_ids[0].tolist():
        if tid != prev:
            collapsed.append(tid)
            prev = tid
    tokens = [
        id2token[tid] for tid in collapsed
        if tid != BLANK_TOKEN_ID and id2token.get(tid, "") not in ("", "$")
    ]
    return " ".join(tokens), confidence


# ── Inference ─────────────────────────────────────────────────
df_test = pd.read_csv(CSV_PATH)
results = []

for _, row in df_test.iterrows():
    sample_id      = row["id"]
    path           = row["path"]
    canonical_text = row["canonical"]
    wav_path       = os.path.join(BASE_DIR, path)

    speech, sr = sf.read(wav_path)
    if len(speech.shape) > 1:
        speech = speech.mean(axis=1)

    trans1, conf1 = get_trans_conf_m1(speech)
    conf2         = get_conf_m2(speech)          # HuBERT: confidence only
    trans3, conf3 = get_trans_conf_m3(speech, canonical_text)

    # Model2 thắng → dùng trans của model có confidence cao nhất trong {1, 3}
    # vì model2 không decode được tiếng Việt
    best_conf = max(conf1, conf2, conf3)

    if best_conf == conf3:
        chosen_trans, chosen_model = trans3, "model3_checkpoint_wl"
    elif best_conf == conf2:
        # HuBERT tự tin nhất → lấy trans của model còn lại tự tin hơn
        fallback_trans, fallback_model = (trans3, "model3_checkpoint_wl(via_m2)") \
            if conf3 >= conf1 else (trans1, "model1_wav2vec2_vi(via_m2)")
        chosen_trans, chosen_model = fallback_trans, fallback_model
    else:
        chosen_trans, chosen_model = trans1, "model1_wav2vec2_vi"

    results.append({
        "id":           sample_id,
        "path":         path,
        "predict":      chosen_trans,
        "chosen_model": chosen_model,
        "conf_m1":      round(conf1, 4),
        "conf_m2":      round(conf2, 4),
        "conf_m3":      round(conf3, 4),
    })

    print(f"[{sample_id}]")
    print(f"  M1 ({conf1:.4f}): {trans1}")
    print(f"  M2 ({conf2:.4f}): [HuBERT - no VI decode]")
    print(f"  M3 ({conf3:.4f}): {trans3}")
    print(f"  → {chosen_model}: {chosen_trans}")
    print("-" * 60)

# ── Xuất kết quả ──────────────────────────────────────────────
result_df = pd.DataFrame(results)
result_df.to_csv(OUTPUT_CSV, index=False, encoding="utf-8")
print(f"Saved to {OUTPUT_CSV}")
```

```python
import pandas as pd
import re
from phonemizer import phonemize

INPUT_PATH = "/home/h/git/MDD-Framework-Public/results.csv"
OUTPUT_PATH = "/home/h/git/MDD-Framework-Public/results_clean_phones.csv"

# load file
df = pd.read_csv(INPUT_PATH)

# 1. chỉ clean khoảng trắng
def clean_spaces(t):
    if pd.isna(t):
        return ""
    t = re.sub(r"\s+", " ", str(t)).strip()
    return t

df["predict"] = df["predict"].apply(clean_spaces)

# 2. convert sang phones (batch = nhanh hơn loop)
phones = phonemize(
    df["predict"].tolist(),
    language="vi",
    backend="espeak",
    strip=True,
    njobs=8  # chỉnh theo CPU
)

df["predict"] = phones

# 3. save file
df.to_csv(OUTPUT_PATH, index=False, encoding="utf-8")

print(f"Saved to {OUTPUT_PATH}")
```
